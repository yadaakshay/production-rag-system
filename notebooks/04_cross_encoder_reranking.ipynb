# Module 4: Precision Boost with Cross-Encoder Reranking

### Bi-Encoders vs Cross-Encoders:
| Feature | Bi-Encoder (Embedding Search) | Cross-Encoder (Reranker) |
|---|---|---|
| Architecture | Query and Passage encoded independently | Full cross-attention over `[CLS] Query [SEP] Passage [SEP]` |
| Speed | Sub-millisecond (ANN vector search over millions) | Slower (requires Transformer forward pass for each pair) |
| Precision | Medium | High (captures intricate token interactions) |
| Best Role | First-stage candidate generation (top 20-50) | Second-stage reranker (top 3-5) |


```python
import sys
sys.path.append('..')

from src.core.models import Chunk, RetrievalResult
from src.retrieval.reranker import CrossEncoderReranker

query = 'How long does the circuit breaker remain open?'

# Candidates retrieved by stage 1
candidates = [
    RetrievalResult(chunk=Chunk(doc_id='1', text='Envoy gateway handles client rate-limiting and mTLS certificates.'), score=0.75, rank=1),
    RetrievalResult(chunk=Chunk(doc_id='2', text='When failure rate exceeds 50%, the circuit breaker trips to OPEN for 30 seconds.'), score=0.72, rank=2),
    RetrievalResult(chunk=Chunk(doc_id='3', text='Redis cache invalidation uses Pub/Sub to purge stale local entries.'), score=0.68, rank=3),
]

reranker = CrossEncoderReranker(model_name='heuristic')
reranked = reranker.rerank(query, candidates, top_n=2)

print('Top candidate after Cross-Encoder reranking:')
print(f'Chunk: {reranked[0].chunk.text}')
print(f'Rerank Score: {reranked[0].score:.4f}')
```

# Module 2: Advanced Chunking Strategies

### Comparing 4 Chunking Methodologies:
1. **Fixed Size**: Naive slicing by character count.
2. **Recursive Character**: Separator hierarchy (`\n\n` -> `\n` -> `. ` -> ` `) preserving paragraph and sentence integrity.
3. **Semantic Chunking**: Dynamic boundary detection using sentence embedding distance thresholds.
4. **Hierarchical (Parent-Child)**: Small child chunks for vector indexing; large parent chunks for LLM context synthesis.


```python
import sys
sys.path.append('..')

from src.core.models import Document
from src.ingestion.chunkers import RecursiveCharacterChunker, HierarchicalChunker

text = '''## Distributed Caching
The platform deploys a two-tier caching hierarchy to protect backend relational datastores.
L1 Cache is an in-process Caffeine cache with a 60-second TTL.

## Messaging Tier
All asynchronous event processing utilizes Apache Kafka with partitioned topics.
Replication factor is set to 3 to guarantee cross-AZ durability.'''

doc = Document(text=text, metadata={'source': 'architecture.md'})

# 1. Recursive Chunking
rec_chunker = RecursiveCharacterChunker(chunk_size=150, chunk_overlap=30)
rec_chunks = rec_chunker.chunk(doc)
print(f'Recursive Chunks generated: {len(rec_chunks)}')
for i, c in enumerate(rec_chunks):
    print(f'-- Chunk {i} ({len(c.text)} chars):\n{c.text}\n')
```

```python
# 2. Hierarchical (Parent-Child) Chunking
hier_chunker = HierarchicalChunker(parent_chunk_size=250, child_chunk_size=80)
child_chunks = hier_chunker.chunk(doc)
print(f'Hierarchical Child Chunks: {len(child_chunks)}')
first_child = child_chunks[0]
print('Child Chunk Text:', first_child.text)
print('Associated Parent Text:', first_child.metadata.get('parent_text'))
```

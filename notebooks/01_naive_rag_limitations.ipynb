### Learning Objectives:
1. Understand the architecture of Naive RAG (Load -> Split -> Vector Embed -> Cosine Top-K -> Prompt).
2. Identify why Naive RAG fails in production on technical and enterprise documents.
3. Witness specific failure modes: exact keyword misses, false positive semantic drift, and chunk boundary truncation.


```python
import sys
from pathlib import Path
sys.path.append('..')

from src.core.models import Document
from src.ingestion.chunkers import FixedSizeChunker
from src.embeddings.api_embedder import MockEmbedder
from src.indexing.vector_store import InMemoryVectorStore

# Ingest a realistic technical document
sample_doc = Document(
    text='Our API gateway enforces asymmetric RS256 JWT tokens. Redis L2 cache uses a 24-hour TTL with idempotency keys to avoid duplicate transaction processing.',
    metadata={'filename': 'arch_spec.md'}
)
print('Original Document:', sample_doc)
```

### Naive Step 1: Fixed Character Chunking
Naive RAG typically splits text at arbitrary fixed intervals (e.g. 50 characters). Observe how this severs critical phrases across boundaries.


```python
chunker = FixedSizeChunker(chunk_size=50, chunk_overlap=10)
chunks = chunker.chunk(sample_doc)
for i, c in enumerate(chunks):
    print(f'Chunk #{i}: "{c.text}"')
```

### Failure Mode Analysis:
- **Boundary Mutilation**: 'asymmetric RS256 JWT' is broken midway, ruining exact term search.
- **Lost Semantic Anchor**: Without paragraph awareness, the context of 'Redis L2 cache' is separated from 'idempotency keys'.

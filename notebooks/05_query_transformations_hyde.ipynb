# Module 5: Query Transformations (HyDE & Query Routing)

### 1. HyDE (Hypothetical Document Embeddings - Gao et al.):
User questions are often short and asymmetric compared to document passages (*'Kafka failover?'* vs 500-word paragraph detailing zookeeper leader election).
HyDE uses an LLM to generate a hypothetical passage that would answer the question, and searches using that passage vector!

### 2. Query Routing:
Not every query requires vector retrieval. Conversational chit-chat (*'Hello!'*) or general arithmetic (*'What is 100 * 5?'*) should be routed to direct response to eliminate latency and avoid retrieving random noise.


```python
import sys
sys.path.append('..')

from src.generation.query_transforms import QueryRouter, HyDETransformer
from src.generation.llm_client import MockLLMClient

llm = MockLLMClient()
router = QueryRouter(llm)

queries = [
    'Hi there, how are you today?',
    'What is the replication factor for partitioned Kafka topics?',
]

for q in queries:
    decision, reason = router.route(q)
    print(f'Query: "{q}" -> Route: {decision} ({reason})')
```

# Module 6: Evaluating RAG with the RAG Triad (Ragas Style)

### The RAG Triad Metrics:
1. **Faithfulness (Anti-Hallucination)**: Is the answer strictly derived from retrieved context?
2. **Answer Relevance**: Does the answer directly address the user query?
3. **Context Precision**: Are the relevant context chunks ranked higher than irrelevant ones?
4. **Context Recall**: Are all key ground-truth facts retrieved?


```python
import sys
sys.path.append('..')

from src.evaluation.metrics import (
    calculate_faithfulness,
    calculate_answer_relevance,
    calculate_context_precision,
    calculate_context_recall,
)

query = 'What is the circuit breaker reset timeout?'
ground_truth = 'The circuit breaker stays in OPEN state for 30 seconds before testing downstream health.'
retrieved_context = ['The circuit breaker monitors call failure rates. When tripped into OPEN state, it waits 30 seconds before transitioning to HALF-OPEN.']
answer = 'According to the documentation, the circuit breaker remains in the OPEN state for 30 seconds before switching to HALF-OPEN.'

faith = calculate_faithfulness(answer, retrieved_context)
relevance = calculate_answer_relevance(query, answer)
precision = calculate_context_precision(ground_truth, retrieved_context)
recall = calculate_context_recall(ground_truth, retrieved_context)

print(f'Faithfulness: {faith * 100:.1f}%')
print(f'Answer Relevance: {relevance * 100:.1f}%')
print(f'Context Precision: {precision * 100:.1f}%')
print(f'Context Recall: {recall * 100:.1f}%')
```
